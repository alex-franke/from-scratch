## NumPy Arrays

```python
# 3 samples, 2 features
X = np.array([[1.0, 2.0], [3.0, 4.0], [-1.0, 0.5]])

w = np.array([0.5, -1.0])
b = 1.0
```

```python
print(X)
print(X.shape)
print(X.dtype)
print(X.ndim)
print("-" * 50)
print(w)
print(w.shape)
```

```python
# Elementwise vs matmul
c = np.array([1, 2])
d = np.array([3, 4])
print(c * d)
print(c @ d)
print(np.dot(c, d))
print(np.matmul(c, d))
```

```python
z = (X @ w) + b
#   (3,)    + () -> scalar b is added to every element of the (3,) array!
print(z)
```

```python
X = np.array([[1.0, 2.0], [3.0, 4.0], [-1.0, 0.5]])
w = np.array([0.5, -1.0])
b = 1.0

scores = X @ w + b
print("Scores:", scores)
print("Shape:", scores.shape)
```
