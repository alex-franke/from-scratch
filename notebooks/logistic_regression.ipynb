# Implementing Sigmoid

```python
import numpy as np

X = np.array(
    [
        [-1.0, 2.0],
        [3.0, -4.0],
        [2.0, 3.0],
    ]
)
w = np.array([1.0, 2.0])
b = 1.0
```

```python
w.shape
```

```python
z = np.matmul(X, w) + b
print(z)
```

```python
positive = z >= 0
result = np.empty_like(z, dtype=float)
```

```python
def sigmoid(x: np.ndarray) -> np.ndarray:
    """Sigmoid implementation with overflow management"""
    positive = x >= 0
    result = np.empty_like(x, dtype=float)
    result[positive] = 1 / (1 + np.exp(-x[positive]))
    result[~positive] = np.exp(x[~positive]) / (1 + np.exp(x[~positive]))
    return result


sigmoid(z)
```

```python
h = np.array([-1000.0, -2.0, 0.0, 2.0, 1000.0])
sigmoid(h)
```

# Probability Prediction Function

```python
def predict_proba(X: np.ndarray, w: np.ndarray, b: float = 0.0) -> np.ndarray:
    return sigmoid(np.matmul(X, w) + b)
```

```python
predict_proba(X, w, b)
```
